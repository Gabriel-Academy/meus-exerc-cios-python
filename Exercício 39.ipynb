```python
n = int(input('Digite o ano do seu nasciumeto: '))
n2 = int(input('Informe o ano atual: '))
n3 = n2 - n
if n3 < 18:
  print(f'Com {n3} anos, você ainda terá que se alistar no Exército! Boa sorte.')
elif n3 == 18:
  print(f'''Com {n3} anos, você já deve se alistar. Procure uma Junta Militar mais proxima
para obter mais informações.''')
else:
  print(f'''Com {n3} anos, você PASSOU do prazo de alistamento, procure uma Junta Militar o mais
rápido possível.''')
```
