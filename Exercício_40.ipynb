```python
n = float(input('Digite dus primeira nota: '))
n2 = float(input('Digite a sua segunda nota: '))
n3 = n + n2
n4 = n3/2
if n4 < 5:
  print(f'Sua nota foi {n4}, portanto você foi REPROVADO.')
elif n4 >= 5 and n4 < 6.9:
  print(f'Sua nota foi {n4}, portanto, você está de RECUPERAÇÃO.')
else:
  print(f'Sua nota foi {n4}, portanto, Você foi APROVADO!')
```
