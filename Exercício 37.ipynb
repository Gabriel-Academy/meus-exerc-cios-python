```python
n = int(input('Digite o valor que você deseja converter: '))
print('''Escolha uma das opções abaixo
[1] BINÁRIO
[2] HEXADECIMAL
[3] OCTAL''')
opção = int(input('Digite uma opção: '))
if opção == 1:
  print(f'O valor convertido é {bin(n)[2:]}')
elif opção == 2:
  print(f'O valor convertido é {hex(n)[2:]}')
elif opção == 3:
  print(f'O valor convertido é {oct(n)[2:]}')
else:
  print('Número inválido, por favor tente novamente.')
```
