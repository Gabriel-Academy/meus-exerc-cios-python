```python
n = float(input('Digite o seu peso: '))
n2 = float(input('Digite sua altura: '))
n3 = n2 * n2
n4 = n / n3
if n4 < 18.5:
  print(f'Com {n4:.2f}, você se enquadra no quadro de ABAIXO DO PESO.')
elif n4 >= 18.5 and n4 <= 25:
  print(f'Com {n4:.2f}, você está no no quadro de PESO IDEAL.')
elif n4 >= 25 and n4 <= 30:
  print(f'Com {n4:.2f}, você está no quadro de SOBREPESO.')
elif n4 >= 30 and n4 <= 40:
  print(f'Com {n4:.2f}, você está no quadro de OBESIDADE.')
else:
  print(f'Com {n4:.2f}, você está no quadro de Obesidade Mórbida.')
```
