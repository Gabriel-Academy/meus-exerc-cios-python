```python
n = float(input('Digite o valor da sua compra: '))
print('''Escolha uma forma de pagamento:
[1] À vista (10%) de desconto
[2] À vista no cartão (5%) de desconto
[3] Em até 2x no cartão
[4] 3x ou mais, no cartão (20%) de juros ''')
n2 = int(input('Selecione a forma de pagamento: '))
n3 = n * 0.10
n4 = n * 0.05
n5 = n * 0.20
if n2 == 1:
  print(f'O valor do seu produto é R$ {n}, à vista ficou R$ {n-n3}.')
elif n2 == 2:
  print(f'O valor do seu prouto é R$ {n}, à vista no cartão ficou R$ {n-n4}')
elif n2 == 3:
  print(f'O valor do seu produto é {n}, em 2x no cartão fica por R$ {n/2}')
if n2 == 4:
  n6 = int(input('Em quantas vezes você deseja parcelar? '))
  print(f'O valor do seu produto é {n}, em {n6}x fica por R$ {n+n5}.')
```
